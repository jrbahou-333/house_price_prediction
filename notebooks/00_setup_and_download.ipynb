## EPC — filtered CSV download

1. Go to <https://get-energy-performance-data.communities.gov.uk/>
2. Sign in / create a free **GOV.UK One Login**.
3. Choose "Download files" and filter to **Sefton** before downloading —
   this gives a ~123 MB CSV, not the 7.5 GB whole-country file.
4. Save it as `data/raw/epc/sefton_certificates.csv`.

(An API also exists — same One Login, a bearer token instead of a file —
useful if a filtered CSV isn't available for a wider area later. Not
needed for Phase 1.)


```python
epc_path = DATA_RAW / "epc" / "sefton_certificates.csv"
print("Expecting:", epc_path)
print(f"Found! ({epc_path.stat().st_size / 1e6:.1f} MB)" if epc_path.exists() else "Not found yet - complete the manual step above.")
```
